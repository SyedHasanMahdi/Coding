```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd


data = pd.read_csv("dataset.csv")
print(data.head())
```

```python
print(data.isnull().sum())
#79 missing values in Age column
data = data.dropna()
```

```python
data["NetWorth"] = data["NetWorth"].str.strip("$") # remove dollar sign
data["NetWorth"] = data["NetWorth"].str.strip("B") # remove Billion sign
data["NetWorth"] = data["NetWorth"].astype(float) #make them into numbers instead of strings
```

```python
df = data.sort_values(by = ["NetWorth"], ascending=False).head(10)
plt.figure(figsize=(20, 10))
sns.histplot(x="Name", hue = "NetWorth", data=df)
plt.show
```

```python
a = data["Source"].value_counts().head()
index = a.index
sources = a.values
plt.figure(figsize = (5,5))
plt.pie(sources, labels=index)

plt.rc('font',size = 12)
plt.title("Top 5 Sectors to become a Billionaire", fontsize =20)
plt.show()
```

```python
a = data["Industry"].value_counts().head()
index = a.index
sources = a.values
plt.figure(figsize = (5,5))
plt.pie(sources, labels=index)

plt.rc('font',size = 12)
plt.title("Top 5 Industries to become a Billionaire", fontsize =20)
plt.show()
```

```python
a = data["Country"].value_counts().head()
index = a.index
Countries = a.values
plt.figure(figsize=(5, 5))
plt.pie(Countries, labels=index)
plt.rc('font', size=12)
plt.title("Top 5 Countries with Most Number of Billionaires", fontsize=20)
plt.show()
```
