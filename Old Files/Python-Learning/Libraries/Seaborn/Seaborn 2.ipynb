Seaborn plotting Functions

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.relplot(data=tips, x="total_bill", y="tip")
```

```python
sns.relplot(data=tips, x="total_bill",y="tip", hue="day")
```

```python
sns.relplot(data=tips, x="total_bill", y="tip", hue="day", col="time", col_wrap=2)
#x,y:specify variables     hue:grouping variable that will make different colors.
#data:input data structure(pandas dataframe, numpy array etc)
#col_wrap: wrap the column variable at this width, so the column spans multiple rows.
```

```python
sns.relplot(data=tips, x="size", y="tip",kind="line",errorbar=None)
#kind: kind of plot to draw, options are scatter and line.
```

```python
df=sns.load_dataset("iris")
sns.histplot(df["petal_length"], kde=False)
```

```python
sns.set_context('paper')
titanic = sns.load_dataset('titanic')
sns.barplot(x='embark_town', y="age", data=titanic, color='purple', errorbar=None)

plt.show()
plt.show()
```

```python
sns.barplot(x='sex',y='survived', hue='class',data=titanic,palette='PuRd', order=['male','female'], capsize=0.05,saturation=0.8,err_kws={'color': 'gray', 'linewidth': 2},errorbar='sd')
# order        → Sets the order of categories along the x-axis
# capsize      → Width of the caps on the error bars (in proportion to bar width)
# saturation   → Controls intensity of bar colors (range: 0 to 1; 1 = full color)
# error bar     → Controls error bar type ('sd' for standard deviation, 'ci' for confidence interval, or None to disable)
# err_kws      → Dictionary for customizing error bar appearance (e.g., {'color': 'gray', 'linewidth': 2})
```

```python
#count plot
sns.set_context('paper')
sns.countplot(x = 'class', hue = 'who', data = titanic, palette = 'deep')
plt.title('Survivors')
plt.show()
```

```python
#Point plot is used to show point estimates and confidence intervals
#A point plot represents an estimate of central tendency for a numeric variable
data = sns.load_dataset("tips")
sns.pointplot(x="time", y="tip", data=data,hue="smoker")
plt.show()
```

```python
sns.pointplot(x="time", y="total_bill", hue="smoker",
             data=data, palette="Accent")
```

```python
#joint plot draws a plot of two variable with graphs of 2 variables(bivariate) and graphs of one variable (univariate)
sns.set_style("dark")
sns.jointplot(x='total_bill', y='tip',data=tips)
```

```python
# Add regression line to scatter plot and kernel density estimate to histogram
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg')
```

```python
# Display kernel density estimate instead of scatter plot and histogram
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde')
```

```python
# Display hexagons instead of points in scatter plot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex')
```
