```python
import time

i = 0
x, y = [], []

fig, ax = plt.subplots()

while i < 50:
    x.append(i)
    y.append(psutil.cpu_percent())

    ax.clear()
    ax.plot(x, y, color='b')

    ax.set_xlim(left=max(0, i - 50), right=i + 10)
    ax.set_ylim(0, 100)
    ax.set_xlabel("Time (iterations)")
    ax.set_ylabel("CPU Usage (%)")
    ax.set_title("Live CPU Usage")

    plt.pause(0.1)  # Pause to update plot
    i += 1

# Save plot
fig.savefig("cpu_usage_plot.png", dpi=300)

plt.show()  # Show plot window and block here
```
