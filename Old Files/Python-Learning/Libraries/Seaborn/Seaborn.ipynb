```python
df = sns.load_dataset("car_crashes")
df.head()
```

Styling and Themes in Seaborn

```python
plt.scatter(df.speeding,df.alcohol)
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set() #we can style this plot using the set() function
plt.show()
```

```python
# seaborn supports various themes. Using set_style() function we can set any of these themes
# some include Darkgrid, Whitegrid, Dark, White,Ticks
plt.scatter(df.speeding,df.alcohol)
sns.set_style("whitegrid")
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set_style("whitegrid")
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set_style("dark")
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set_style("white")
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set_style("ticks")
plt.show()
```

```python
#usually we don't need the top and right axis spines so we can use despine() to get rid of them
plt.scatter(df.speeding,df.alcohol)
sns.set_style("ticks")
sns.despine()
plt.show()
```

```python
#styles can be customized by adding a parameter to the set_style() function
#Parameters available are viewed by axes_style()
param=sns.axes_style()
param
```

```python
plt.scatter('speeding','alcohol', data=df)
sns.set_style("darkgrid", {'grid_color':'.5'})
sns.despine()
plt.show()
```

```python
#we can control individual elements and control the scale of the elements or the plot by using set_context().
#there are 4 preset templates. Paper, Notebook, Talk, Poster
plt.scatter(df.speeding, df.alcohol)
sns.set_style("dark")
sns.set_context("notebook")
plt.show()
```

```python
plt.scatter(df.speeding,df.alcohol)
sns.set_style("dark")
sns.set_context("poster")
plt.show()
```
