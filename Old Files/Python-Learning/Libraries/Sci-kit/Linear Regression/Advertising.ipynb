```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import seaborn as sns
```

```python
df = pd.read_csv("Advertising.csv")
advertising_data = df.loc[:,['TV', 'radio', 'newspaper', 'sales']]
```

```python
sales = df.loc[:, 'sales']
tv = df.loc[:, 'TV']
radio = df.loc[:, 'radio']
newspaper = df.loc[:, 'newspaper']
```

```python
fig = plt.figure()
plt.subplot(3,3,1)
plt.scatter(tv,sales, s=10, color='blue')
plt.subplot(3,3,2)
plt.scatter(radio,sales, s=10, color = 'red')
plt.subplot(3,3,3)
plt.scatter(newspaper,sales, s=10, color = 'green')
```

```python
model = LinearRegression()
x = df.loc[:, 'TV'].values
y = df.loc[:, 'sales'].values
x=x.reshape(-1, 1)
model.fit(x,y)
y_pred = model.predict(x)
```
